```python
from numba import cuda

def get_gpu_info():
    print("=== Labwork 2: Get to know your GPU ===")
    
    print("\nDetecting GPUs...")
    cuda.detect()

    if not cuda.is_available():
        print("No CUDA-capable GPU found.")
        return

    device_id = 0
    device = cuda.select_device(device_id)
    print(f"\n--- Selected GPU ID: {device_id} ---")

    print(f"Device Name: {device.name.decode('utf-8')}")
    
    sm_count = getattr(device, 'MULTIPROCESSOR_COUNT')
    cc = getattr(device, 'COMPUTE_CAPABILITY') 
    
    cores_per_sm_map = {
        (5, 0): 128, (5, 2): 128, (5, 3): 128, 
        (6, 0): 64,  (6, 1): 128, (6, 2): 128, 
        (7, 0): 64,  (7, 2): 64,  (7, 5): 64,  
        (8, 0): 64,  (8, 6): 128, (8, 9): 128, 
        (9, 0): 128                            
    }
    
    cores_per_sm = cores_per_sm_map.get(cc, 64) 
    total_cores = sm_count * cores_per_sm
    
    print(f"Compute Capability: {cc[0]}.{cc[1]}")
    print(f"Multiprocessor Count (SMs): {sm_count}")
    print(f"Total CUDA Cores: {total_cores} ({cores_per_sm} cores/SM)")
    
    context = cuda.current_context()
    free_mem, total_mem = context.get_memory_info()

    total_mem_gb = total_mem / (1024 ** 3)
    free_mem_gb = free_mem / (1024 ** 3)
    
    print(f"Total Global Memory: {total_mem_gb:.2f} GB")
    print(f"Free Global Memory:  {free_mem_gb:.2f} GB")

if __name__ == '__main__':
    get_gpu_info()
```
